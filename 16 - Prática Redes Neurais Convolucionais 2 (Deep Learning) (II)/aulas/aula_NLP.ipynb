```python
!pip install -q tensorflow
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
import numpy as np
import matplotlib.pyplot as plt
```

```python
setences = [
    "sequencia de texto inicial",
    "Gosto de sanduiche de milho",
    "Talvez eu compre um celta"
]
```

```python
MAX_VOCAB_SIZE = 20000
tokenizer = Tokenizer(num_words=MAX_VOCAB_SIZE) #instancia de Tokenizer com 20000 palavras
tokenizer.fit_on_texts(setences) # aprende as sentenças
sequences = tokenizer.texts_to_sequences(setences)
```

```python
print(sequences)
```

```python
tokenizer.word_index # elemento 0 é para padding
```

```python
data = pad_sequences(sequences, maxlen=5)
print(data)
```

```python
MAX_SEQUENCE_LENGTH = 5
data = pad_sequences(sequences, maxlen=MAX_SEQUENCE_LENGTH)
print(data)
```

```python
data = pad_sequences(sequences, maxlen=MAX_SEQUENCE_LENGTH, padding='post') # padding 0 ao final
print(data)
```

```python
data = pad_sequences(sequences, maxlen=10) # muito padding
print(data)
```

```python
data = pad_sequences(sequences, maxlen=3) # pouco padding trunca
print(data)
```

```python
data = pad_sequences(sequences, maxlen=3, truncating='post') # pouco padding trunca
print(data)
```
