```python
!pip install -q tensorflow
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Embedding, GlobalAveragePooling1D, Dense
from tensorflow.keras.models import Model
import numpy as np
```

```python

# avaliacoes de filmes
sentences = [
    "Eu amei esse filme",
    "O filme foi muito bom",
    "Que horror de filme",
    "Muito ruim, não gostei",
    "Espetacular! Nota 10",
    "Odiei, nota 10 kkkk" # ruido
]

# 1 = positivo, 0 = negativo
labels = np.array([1, 1, 0, 0, 1, 1])

# tokenização
MAX_VOCAB = 1000
tokenizer = Tokenizer(num_words=MAX_VOCAB, oov_token="<OOV>") # oov é Out Of Vocabulary ou seja se uma palavra não estiver entre as 1000 mais comuns aprendidas durante o fit_on_texts() substitua pelo token <OOV>
tokenizer.fit_on_texts(sentences)

sequences = tokenizer.texts_to_sequences(sentences)
padded = pad_sequences(sequences, padding='post')

# modelo
# tamanho indefinido de sequência de inteiros
inputs = Input(shape=(None,))

# API functional
x = Embedding(MAX_VOCAB, 16)(inputs) # camada dos embberdings
x = GlobalAveragePooling1D()(x) # pooling com media
x = Dense(16, activation='relu')(x) # camada oculta
outputs = Dense(1, activation='sigmoid')(x)

model = Model(inputs=inputs, outputs=outputs)
model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])

model.fit(padded, labels, epochs=30, verbose=1)

# teste com frase nova
nova_frase = ["Esse filme foi bom demais"]
nova_seq = tokenizer.texts_to_sequences(nova_frase)
nova_pad = pad_sequences(nova_seq, maxlen=padded.shape[1], padding='post')

# predição
pred = model.predict(nova_pad)
print("probabilidade de ser positivo:", pred[0][0]) #treinamento não muito bom, base pequena demais
```
