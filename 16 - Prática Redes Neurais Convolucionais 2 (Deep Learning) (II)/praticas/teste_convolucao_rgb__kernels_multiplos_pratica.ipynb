
# convolução em imagem colorida



```python
import torch.nn as nn
import torch

#imagem RGB com 32x32 pixels
imagem = torch.randn(1, 3, 32, 32)  # batch_size, canais rgb, altura, largura
print("Dimensões da entrada")
print(imagem.shape)

# kernel
conv = nn.Conv2d(in_channels=3, out_channels=16, kernel_size=3, padding=1)

# convolução
saida = conv(imagem)
print("Dimensões da saída")
print(saida.shape)
```
